```python
# Define the feature and target variables
# X = df[["RM"]]  # Using only the number of rooms as a feature
X = df.drop("MEDV", axis=1)  # Using all features except the target variable
y = df["MEDV"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.15,
    random_state=42
)

print("Training samples:", len(X_train))
print("Test samples:", len(X_test))
```

```python
from sklearn.linear_model import LinearRegression
```

```python
# Initialize and train the model
model = LinearRegression()
model.fit(X_train, y_train)

print("Model training completed.")
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
# Generate predictions on the test set
y_pred = model.predict(X_test)

# Compare actual and predicted values
results = pd.DataFrame({
    "Actual Value": y_test,
    "Predicted Value": y_pred
})

display(results.head(10))
```

```python
# Visualize actual vs. predicted values
plt.figure(figsize=(8, 5))

plt.scatter(y_test, y_pred, alpha=0.6)

plt.xlabel("Actual Home Value ($1,000s)")
plt.ylabel("Predicted Home Value ($1,000s)")
plt.title("Actual vs. Predicted Home Values")

# Add a reference line for perfect predictions
min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# Calculate Mean Absolute Error
mae = mean_absolute_error(y_test, y_pred)

# Calculate Root Mean Squared Error
rmse = root_mean_squared_error(y_test, y_pred)

print(f"Mean Absolute Error: {mae:.3f} ($1,000s)")
print(f"Root Mean Squared Error: {rmse:.3f} ($1,000s)")
```

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# Create and train a baseline model
baseline_model = DummyRegressor(strategy="median")
baseline_model.fit(X_train, y_train)

# Generate baseline predictions
baseline_pred = baseline_model.predict(X_test)

# Evaluate both models
baseline_mae = mean_absolute_error(y_test, baseline_pred)
baseline_rmse = root_mean_squared_error(y_test, baseline_pred)

print(f"Baseline MAE: {baseline_mae:.3f} ($1,000s)")
print(f"Linear Regression MAE: {mae:.3f} ($1,000s)")
print(f"Baseline RMSE: {baseline_rmse:.3f} ($1,000s)")
print(f"Linear Regression RMSE: {rmse:.3f} ($1,000s)")
```

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, cross_val_score
import pandas as pd

# Define cross-validation
cv = KFold(n_splits=5, shuffle=True, random_state=42)

results = []

# Evaluate models with different numbers of features
for n_features in range(3, X_train.shape[1]):
    selector = SequentialFeatureSelector(
        estimator=LinearRegression(),
        n_features_to_select=n_features,
        direction="forward",
        scoring="neg_mean_absolute_error",
        cv=cv,
        n_jobs=-1
    )

    selector.fit(X_train, y_train)

    selected_columns = X_train.columns[selector.get_support()]
    X_subset = X_train[selected_columns]

    scores = cross_val_score(
        LinearRegression(),
        X_subset,
        y_train,
        scoring="neg_mean_absolute_error",
        cv=cv,
        n_jobs=-1
    )

    results.append({
        "Number of Features": n_features,
        "Cross-Validation MAE": -scores.mean()
    })

results_df = pd.DataFrame(results)
display(results_df.sort_values("Cross-Validation MAE"))
```

```python

# Select the 9 best features using forward selection
selector_9 = SequentialFeatureSelector(
    estimator=LinearRegression(),
    n_features_to_select=9,
    direction="forward",
    scoring="neg_mean_absolute_error",
    cv=cv,
    n_jobs=-1
)

# Fit the selector using training data only
selector_9.fit(X_train, y_train)

# Store the selected feature names
selected_features = X_train.columns[
    selector_9.get_support()
].tolist()

print("Selected features:")
print(selected_features)
```

```python
# Prepare the reduced feature datasets
X_train_selected = X_train[selected_features]
X_test_selected = X_test[selected_features]

# Train the reduced model
selected_model = LinearRegression()
selected_model.fit(X_train_selected, y_train)

# Generate predictions
y_pred_selected = selected_model.predict(X_test_selected)
```

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# Evaluate the selected-feature model
mae_selected = mean_absolute_error(y_test, y_pred_selected)
rmse_selected = root_mean_squared_error(y_test, y_pred_selected)

print(f"All-features MAE: {mae:.3f} ($1,000s)")
print(f"Selected-features MAE: {mae_selected:.3f} ($1,000s)")
print(f"All-features RMSE: {rmse:.3f} ($1,000s)")
print(f"Selected-features RMSE: {rmse_selected:.3f} ($1,000s)")
```

```python

# Select one existing house from the dataset
sample_house = X_test.iloc[[0]]

# Get its actual value
actual_value = y_test.iloc[0]

# Predict its value using the 9-feature model
predicted_value = selected_model.predict(
    sample_house[selected_features]
)[0]

print(f"Actual value: ${actual_value * 1000:,.0f}")
print(f"Predicted value: ${predicted_value * 1000:,.0f}")
print(
    f"Absolute error: "
    f"${abs(actual_value - predicted_value) * 1000:,.0f}"
)
```
